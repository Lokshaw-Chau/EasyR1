```python
import datasets

dataset = datasets.load_dataset("parquet", data_files="/root/workspace/datasets/gui-r1/uground_bias_p15_sampled_3072.parquet", split="train")

print(dataset[0].keys())
```

```python
for item in dataset:
    print(item['instruction'])
```

```python
cnt = 0
for item in dataset:
    # print(item['instruction'])
    if ',' in item['instruction']:
        cnt += 1
print(f"Total items with comma in instruction: {cnt}")
```

```python
cnt = 0
hint_token = ['above', 'below', 'left', 'right', 'top', 'bottom']
for item in dataset:
    # print(item['instruction'])
    if any(token in item['instruction'] for token in hint_token):
        cnt += 1
print(f"Total items with hint token in instruction: {cnt}")
```

```python
instruction = 'View all posts in News, to the left of "Gunmen Kill Policeman In Rivers State"'
instruction = instruction.split(',')
print(instruction)
for sub_instruction in instruction:
    if any(token in sub_instruction for token in hint_token):
        instruction.remove(sub_instruction)
instruction = ','.join(instruction)
print(instruction)
```

```python
cnt = 0
hint_token = ['above', 'below', 'left', 'right', 'top', 'bottom']
new_dataset = []
for item in dataset:
    instruction = item['instruction']
    instruction = instruction.split(',')
    for sub_instruction in instruction:
        if any(token in sub_instruction for token in hint_token):
            instruction.remove(sub_instruction)
    if len(instruction) == 0:
        cnt += 1
    else:
        new_instruction = ','.join(instruction)
        item['instruction'] = new_instruction
        new_dataset.append(item)

print(f"Total items with no instruction after removing hint tokens: {cnt}")
```

```python
print(len(new_dataset))
```

```python
for item in new_dataset:
    print(item['instruction'])
```

```python
# Save the new dataset to a parquet file
new_dataset = datasets.Dataset.from_list(new_dataset)
new_dataset.to_parquet("/root/workspace/datasets/gui-r1/uground_bias_p15_sampled_3041_no_hint.parquet")
```
